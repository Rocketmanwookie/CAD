```python
import ipywidgets as widgets
from IPython.display import display

checks = [
    Check('F-01', 'Link the candidate workbench, start FreeCAD, and select Controls / Automation.'),
    Check('F-02', 'Create one project and record project/intake and typed I/O data.'),
    Check('F-03', 'Allocate catalog I/O and verify capacity, persistent rack/module/channel occurrences, intended spare behavior, and a clean Report View.'),
    Check('F-04', 'Create an electrical path from an allocated channel to a terminal and field-device endpoint.'),
    Check('F-05', 'Save the FCStd inside the project root, close it, reopen it, and recompute without lifecycle errors.'),
    Check('F-06', 'Verify CE identities, path/signal links, allocation links, and the PLC-terminal address invariant after reopen/recompute.'),
    Check('F-07', 'Edit a path and confirm the first PLC terminal cannot change away from its allocated address.'),
    Check('F-08', 'Run project validation and record all findings, including an explicit no-unexpected-findings result if applicable.'),
    Check('F-09', 'Export wiring and I/O-path schedules; retain generated CSVs under the project root and compare them to the persisted path.'),
    Check('F-10', 'Record FreeCAD version, candidate commit, screenshots, Report View output, FCStd, and generated files.'),
]

STATUS_OPTIONS = ('NOT RUN', 'PASS', 'FAIL', 'BLOCKED', 'DEFERRED')
operator_controls = {}
rows = []
for check in checks:
    status = widgets.Dropdown(
        options=STATUS_OPTIONS, value=check.result, description=f'{check.id} status:',
        layout=widgets.Layout(width='260px'),
    )
    notes = widgets.Textarea(
        placeholder='What happened? For FAIL/BLOCKED, paste the exact Report View or error text.',
        layout=widgets.Layout(width='720px', height='72px'),
    )
    evidence = widgets.Text(
        placeholder='screenshots/F-03.png, report-view/F-03.txt',
        description='Evidence:', layout=widgets.Layout(width='720px'),
    )
    operator_controls[check.id] = {'status': status, 'notes': notes, 'evidence': evidence}
    rows.append(widgets.VBox([
        widgets.HTML(f'<b>{check.id}</b> — {check.instruction}'),
        status, notes, evidence,
    ], layout=widgets.Layout(border='1px solid #cccccc', padding='8px', margin='4px 0')))

display(widgets.VBox(rows))
print('Select a status for every check, enter notes/evidence, then run the next cell to collect your entries.')
```

```python
# Run this after making selections above. Evidence paths may be separated by commas or new lines.
results = {}
for check in checks:
    controls = operator_controls[check.id]
    evidence_paths = [
        item.strip() for item in controls['evidence'].value.replace('\n', ',').split(',')
        if item.strip()
    ]
    results[check.id] = {
        'result': controls['status'].value,
        'notes': controls['notes'].value.strip(),
        'evidence': evidence_paths,
    }
    check.result = results[check.id]['result']
    check.notes = results[check.id]['notes']
    check.evidence = results[check.id]['evidence']

summary = {status: sum(check.result == status for check in checks) for status in STATUS_OPTIONS}
summary
```
